```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns
#plotly for  3d graphs
import plotly.graph_objects as go
from sklearn.cluster import KMeans
from sklearn.preprocessing import LabelEncoder
```

```python
#read the data
df=pd.read_csv("/content/customers.csv")
df.head()
```

```python
df=df.drop("CustomerID",axis=1)
```

```python
lb=LabelEncoder()
df['Gender']=lb.fit_transform(df['Gender'])
```

```python
df.head()
```

```python
#visualize the data
plt.figure(figsize=(15,6))
columns=["Age","Annual Income (k$)","Spending Score (1-100)"]
for position,column in enumerate(columns):
  plt.subplot(1,3,position+1)
  sns.histplot(data=df,x=column,kde=True)
  plt.title(f"Distribution of {column}")
plt.tight_layout()#spacing
plt.show()
```

```python
#compare the features
sns.pairplot(df,vars=["Age","Annual Income (k$)","Spending Score (1-100)"],hue='Gender')
```

```python
df.head(20)
```

```python
#select the feature
x1=df[["Age","Annual Income (k$)","Spending Score (1-100)"]].values
#empty list-store how the values change
inertia=[]
#elbow method-run an algorithm at different ks to determine the best k
for n in range(1,16):
  model=KMeans(n_clusters=n,init="k-means++",n_init=10,max_iter=300,tol=0.0001,random_state=111)
  model.fit(x1)
  #store the model change
  inertia.append(model.inertia_)
```

```python
#determine the optimum k
plt.figure(figsize=(15,6))
plt.plot(range(1,16),inertia,marker="o")
plt.title("Determining the optimum k using elbow method")
plt.xlabel("Number of ks")
plt.ylabel("Inertia")
plt.show()
```

```python
model1=KMeans(n_clusters=6,init="k-means++",n_init=10,max_iter=300,tol=0.0001,random_state=111)
model1.fit(x1)
#get clusters
labels=model1.labels_
centroids=model1.cluster_centers_
```

```python
#visualize the clustering
plt.figure(figsize=(15,6))
plt.scatter(x1[:,0],x1[:,1],c=labels,s=100,cmap="rainbow")#original data
#clusters
plt.scatter(centroids[:,0],centroids[:,1],c="yellow",s=300,marker="X",label="Centroids")
plt.title("Clustering based on Spending Score")
plt.xlabel("Spending Score")
plt.ylabel("Clustering")
plt.show()
```
